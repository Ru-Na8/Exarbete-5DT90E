## 1. OMA Heatmaps
Heatmaps show the expert consensus scores (Gem_Avvikelse and Gem_Säker) per participant and movement phase.

- **Gem_Avvikelse** (0-2): deviation from normative movement — higher = worse
- **Gem_Säker** (0-2): rater certainty — higher = more confident

```python
HEATMAP_DIR = Path('../data/processed/OMA_Score/Heatmaps')

print('=== Sit-to-Stand: Avvikelse ===')
display(Image(str(HEATMAP_DIR / 'heatmap_avvikelse_kors.png'), width=800))

print('=== Stand-to-Sit: Avvikelse ===')
display(Image(str(HEATMAP_DIR / 'heatmap_avvikelse_rak.png'), width=800))
```

```python
print('=== Sit-to-Stand: Säker ===')
display(Image(str(HEATMAP_DIR / 'heatmap_saker_kors.png'), width=800))

print('=== Stand-to-Sit: Säker ===')
display(Image(str(HEATMAP_DIR / 'heatmap_saker_rak.png'), width=800))
```

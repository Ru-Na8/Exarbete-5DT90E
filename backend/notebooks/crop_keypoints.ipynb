# Crop keypoint csv files

* Only use times from the excel-files
* Use float times rather than integers

```python
from pathlib import Path
import pandas as pd

KEYPOINT_DIR = Path('../data/processed/keypoints')
OUT_DIR = Path('../data/processed/keypoints_cropped')

time_dict = {
  '028': (9.90, 11.77),
  '030': (5.80, 7.63),
  '045': (6.54, 7.98),
  '047': (7.03, 8.91),
  '059': (6.65, 9.11),
  '061': (6.00, 9.00),
  '074': (6.07, 7.27),
  '076': (5.00, 8.00),
  '091': (6.39, 7.61),
  '093': (5.86, 7.67)
}

stand_to_sit_time_dict = {
  '028': (11.87, 12.87),
  '030': (8.07, 11.56),
  '045': (7.94, 9.26),
  '047': (9.20, 12.30),
  '059': (9.11, 11.06),
  '061': (10.57, 13.41),
  '074': (7.27, 8.20),
  '076': (8.27, 10.84),
  '091': (7.56, 8.18),
  '093': (8.50, 11.19)
}


def crop_keypoints(df, start, end, fps=50):
  df['time'] = df['frame'] / fps
  cropped_df = df[(df['time'] >= start) & (df['time'] <= end)]
  return cropped_df


for model in ['movenet', 'mediapipe_norm', 'mediapipe_world']:
  for vid_id in time_dict.keys():
    file = [file for file in Path(KEYPOINT_DIR / model).iterdir() if str(file).endswith(f'{vid_id}_D_{model}.csv')][0]
    df = pd.read_csv(file)
    times = time_dict[vid_id]
    cropped_df = crop_keypoints(df, *times)

    out_dir = OUT_DIR / model
    Path(out_dir).mkdir(parents=True, exist_ok=True)
    out_path = out_dir / file.name
    cropped_df.to_csv(out_path, sep=',', index=False)
    print(f'{model} {vid_id}: {len(cropped_df)} frames')


DOWN_OUT_DIR = Path('../data/processed/keypoints_cropped_down')

for model in ['movenet', 'mediapipe_norm', 'mediapipe_world']:
  for vid_id in stand_to_sit_time_dict.keys():
    file = [file for file in Path(KEYPOINT_DIR / model).iterdir() if str(file).endswith(f'{vid_id}_D_{model}.csv')][0]
    df = pd.read_csv(file)
    times = stand_to_sit_time_dict[vid_id]
    cropped_df = crop_keypoints(df, *times)

    out_dir = DOWN_OUT_DIR / model
    Path(out_dir).mkdir(parents=True, exist_ok=True)
    out_path = out_dir / file.name
    cropped_df.to_csv(out_path, sep=',', index=False)
    print(f'{model} {vid_id}: {len(cropped_df)} frames')
```
